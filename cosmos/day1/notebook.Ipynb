day 1 of class

```python
class Calculate:
    def __init__(self , num1 , num2):
        self.a = num1
        self.b = num2
        self.total1 = 0

    def total(self):
        self.total1 = self.a + self.b
        return self.total1
    
    def average(self):
        av = (self.total1/2.0)
        return av

num1 = float(input("Enter first: "))
num2 = float(input("Enter second: "))

cal = Calculate(num1 , num2)

print(f"Total: {cal.total()}")
print(f"Average: {cal.average()}")
```

Looping and Condition

```python
import random as r
marks = []
passs = []
fail = []
i = 0
while i < 10:
    marks.append(r.randint(1, 101))
    i += 1
i = 0
while i< 10:
    if marks[i]<50:
        fail.append(marks[i])
    else:
        passs.append(marks[i])
    i += 1


print(f" Passed : {passs}")
print(f" Failed : {fail}")
```

```python
n = int(input("Enter a number: " ))
def check_even_odd(n: int)->None:
    if n %2 ==0 :
        print("Even")
    else:
        print("Odd")

check_even_odd(n)
```
