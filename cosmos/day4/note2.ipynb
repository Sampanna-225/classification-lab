QN Write a programme using lambda function if marks is greater than 80 then it is Distinction and if marks is greater than 60 it is called first division and if the marks is greater than 40 it is second division. then fail. My first marks is 90.

```python
a = int(input("Enter a number"))

result = lambda x: "Distinction" if x>80 and x<=100 else "First Division" if x>60 and x<=80 else "Second Division" if x>40 and x<=60 else "Fail" if x>=0 and x<=40 else "Invalid Input"

print(result(a))
```
