Object Orented Programming Language

```python
class Calculator:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def add(self):
        return self.x + self.y

no = Calculator(1 , 2)
print(no.add())
```

```python
class Person:
    def __init__(self , name , age , height):
        self.a = name
        self.b = age
        self.c = height

    def display(self):
        print(f"Hello my name is {self.a} , my age is {self.b} and height is {self.c}cm.")

name = "Sampanna Adhikari"
age = 18
height = 180

p = Person(name , age , height) 
p.display()
```

```python
class Car:
    def __init__(self, name , engine , brand , color , price):
        self.a = name
        self.b = engine
        self.c = brand
        self.d = color
        self.e = price

    def display(self):
        print(f"Car : {self.c}\nEngine: {self.b} \nBrand : {self.c}\n")

    def show(self):
        print(f"A {self.d} colored {self.c} {self.a} has entered the showroom")

    def calculate(self):
        ratio = self.e / 25745.0
        try:
            print(f"This car is Approx {ratio:.2f} X The price of a honda sivic in $")
        except Exception as e:
            print(e)

name = input("Car Name : ")
engine = input("Car Engine Type : ")
brand = input("Car Brand: ")
color = input("Car Color : ")
price = float(input("Enter Price : ")) 
price = price if price> 0.0 else 1

c = Car(name , engine , brand , color , price)
c.display()
c.show()
c.calculate()
```
