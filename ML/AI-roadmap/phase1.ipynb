```python
    import random
def process_sensor_stream(x:list)->dict:
    try:
        valid_reading = [val for val in x if 0.0 <= val <=100.0]
        x = valid_reading
        if not valid_reading:
            return {"count": 0, "mean": 0.0, "max": 0.0, "min": 0.0}

        count = len(x)
        mean = sum(x)/float(count)
        max_val = max(x)
        min_val = min(x)
        return {
            "count" : count ,
            "mean" : mean,
            "max" : max_val,
            "min" : min_val,
        }
        
        

    except Exception as e:
         print(f"Fatal Error {e} ")


sensor_list = []

for i in range(10): 
    sensor_list.append(random.uniform(-10.0 , 150.0)) #to get randomly generated floating number

result = process_sensor_stream(sensor_list)
print(f"Count : {result['count']}\nMean : {result['mean']:.2f}\nMax : {result['max']:.2f}\nMin : {result['min']:.2f}")
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error , r2_score


X = np.array([[3.5,1.2],[5.0,2.0],[7.2,3.1],[9.0,4.0],[11.5,4.8]])
Y = np.array([[18.3],[25.0],[36.8],[46.5],[57.2]])

predML = LinearRegression()
predML.fit(X , Y)
y_pred = predML.predict(np.array([[5.2,1.8],[3.5,1.2]]))
test = np.array([[26.8],[18.3]])
mse = mean_squared_error(test , y_pred )
r2 = r2_score(test , y_pred)#requires multiple samples to calculate variance 
print(r2)
```
